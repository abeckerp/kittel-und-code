## ⭐ 1. NumPy importieren

Per Konvention wird NumPy immer als `np` abgekürzt. Das sehen Sie in praktisch jedem Code im Internet.

````python
import numpy as np
````

## ⭐ 2. Warum nicht einfach Python-Listen?

Stellen Sie sich vor, Sie haben systolische Blutdruckwerte (in mmHg) von sechs Patient\*innen und möchten jeden Wert um 10 % erhöhen. Mit einer normalen Python-Liste klappt das nicht so, wie man erwarten würde:

````python
blutdruck_liste = [120, 135, 142, 118, 160, 128]

print(blutdruck_liste * 2)      # Die Liste wird verdoppelt, nicht die Werte!
````

Mit einem NumPy-Array wird die Rechnung dagegen **auf jedes Element** angewendet. Das nennt man *Vektorisierung* – und es ist nicht nur bequemer, sondern bei großen Datenmengen auch viel schneller.

````python
blutdruck = np.array([120, 135, 142, 118, 160, 128])

print(blutdruck * 2)
print(blutdruck * 1.1)
````

## ⭐ 3. Arrays erstellen und untersuchen

Jedes Array hat ein paar wichtige Eigenschaften:

| Eigenschaft | Bedeutung |
|---|---|
| `shape` | Form des Arrays, z. B. `(6,)` = 6 Werte in einer Reihe |
| `size` | Anzahl aller Elemente |
| `dtype` | Datentyp, z. B. `int64` (ganze Zahlen) oder `float64` (Kommazahlen) |
| `ndim` | Anzahl der Dimensionen (1 = Liste, 2 = Tabelle) |

````python
print("shape:", blutdruck.shape)
print("size: ", blutdruck.size)
print("dtype:", blutdruck.dtype)
print("ndim: ", blutdruck.ndim)
````

Neben `np.array(...)` gibt es praktische Funktionen, um Arrays zu erzeugen:

````python
tage = np.arange(1, 8)              # 1 bis 7 (die 8 ist NICHT dabei!)
print("Tage:", tage)

zeitpunkte = np.linspace(0, 24, 5)  # 5 gleichmäßig verteilte Werte von 0 bis 24
print("Messzeitpunkte (h):", zeitpunkte)

leer = np.zeros(4)                  # 4 Nullen, z. B. als Platzhalter
print("Nullen:", leer)
````

## ⭐ 4. Rechnen mit ganzen Messreihen

Ein klassisches Beispiel aus dem Klinikalltag: **Blutzucker** wird in Deutschland oft in mg/dl angegeben, international in mmol/l. Umrechnung: mg/dl ÷ 18 ≈ mmol/l.

````python
glukose_mg_dl = np.array([85, 110, 126, 98, 200, 145, 72])

glukose_mmol_l = glukose_mg_dl / 18
print(np.round(glukose_mmol_l, 1))
````

Auch Rechnungen **zwischen** Arrays funktionieren elementweise. Beispiel: Body-Mass-Index

$$\text{BMI} = \frac{\text{Gewicht [kg]}}{\text{Größe [m]}^2}$$

````python
gewicht_kg = np.array([70, 85, 62, 110, 54])
groesse_m  = np.array([1.75, 1.80, 1.65, 1.78, 1.60])

bmi = gewicht_kg / groesse_m**2
print(np.round(bmi, 1))
````

> **Wichtig:** Beide Arrays müssen gleich lang sein. Der erste BMI-Wert gehört zum ersten Gewicht und zur ersten Größe usw.

## ⭐ 5. Deskriptive Statistik

NumPy bringt alle grundlegenden Kennzahlen mit:

````python
print(f"Mittelwert:         {np.mean(blutdruck):.1f} mmHg")
print(f"Median:             {np.median(blutdruck):.1f} mmHg")
print(f"Standardabweichung: {np.std(blutdruck, ddof=1):.1f} mmHg")
print(f"Minimum:            {np.min(blutdruck)} mmHg")
print(f"Maximum:            {np.max(blutdruck)} mmHg")
print(f"Index des Maximums: {np.argmax(blutdruck)}")
````

> **Zu `ddof=1`:** Ohne diesen Parameter berechnet NumPy die Standardabweichung der *Grundgesamtheit* (Teilen durch *n*). Mit `ddof=1` erhalten Sie die *Stichproben*-Standardabweichung (Teilen durch *n − 1*), wie sie in Statistikprogrammen und Publikationen üblich ist.

Viele Funktionen gibt es auch als „Methode“ direkt am Array – das Ergebnis ist dasselbe:

````python
print(blutdruck.mean(), blutdruck.max())
````

## ⭐ 6. Filtern mit Bedingungen (Boolean Indexing)

Das ist eine der nützlichsten Techniken überhaupt. Ein Vergleich mit einem Array liefert ein Array aus `True`/`False`:

````python
hypertonie = blutdruck >= 140
print(hypertonie)
````

Dieses „Masken-Array“ kann man in eckige Klammern stecken, um nur die passenden Werte zu erhalten:

````python
print("Werte mit Hypertonie:", blutdruck[hypertonie])

# Kurzform – das sieht man meistens:
print("Werte unter 130:     ", blutdruck[blutdruck < 130])
````

Ein praktischer Trick: `True` zählt als 1 und `False` als 0. Damit lassen sich Anzahlen und Anteile direkt berechnen:

````python
print("Anzahl mit Hypertonie:", hypertonie.sum())
print(f"Anteil mit Hypertonie: {hypertonie.mean():.0%}")
````

Mehrere Bedingungen kombinieren Sie mit `&` (und) bzw. `|` (oder). **Klammern um jede Bedingung sind Pflicht!**

````python
# Werte von 120 bis 139 mmHg (normal bis hochnormal)
print(blutdruck[(blutdruck >= 120) & (blutdruck < 140)])
````

## 7. Indizieren und Slicen

Einzelne Werte oder Bereiche holen Sie mit eckigen Klammern. **Achtung:** Python beginnt bei **0** zu zählen!

````python
print(blutdruck)
print("Erster Wert:     ", blutdruck[0])
print("Letzter Wert:    ", blutdruck[-1])
print("Werte 2 bis 4:   ", blutdruck[1:4])   # Index 1, 2, 3 – die 4 ist exklusiv
print("Die ersten drei: ", blutdruck[:3])
````

## ⭐ 8. Zweidimensionale Arrays: Patient\*innen × Messgrößen

Echte Daten sind meist Tabellen. In einem 2D-Array entspricht jede **Zeile** einer Patientin / einem Patienten und jede **Spalte** einer Messgröße.

````python
# Spalten: systolisch, diastolisch, Puls
vitalwerte = np.array([
    [120, 80, 72],
    [135, 85, 68],
    [142, 92, 80],
    [118, 76, 65],
    [160, 98, 88],
])

print(vitalwerte.shape)   # (5 Zeilen, 3 Spalten)
````

Zugriff erfolgt mit `[zeile, spalte]`. Ein Doppelpunkt `:` bedeutet „alle“:

````python
print("Patient 3, alle Werte:    ", vitalwerte[2])
print("Alle Patienten, Puls:     ", vitalwerte[:, 2])
print("Patient 1, diastolisch:   ", vitalwerte[0, 1])
````

### Der `axis`-Parameter

Bei Statistiken auf 2D-Arrays müssen Sie sagen, *in welche Richtung* gerechnet wird:

- `axis=0` → **entlang der Zeilen nach unten** → ein Ergebnis pro **Spalte** (z. B. mittlerer Puls aller Patienten)
- `axis=1` → **entlang der Spalten nach rechts** → ein Ergebnis pro **Zeile** (z. B. Mittelwert pro Patient)

```
            axis=1 →
          sys  dia  puls
axis=0  [ 120   80   72 ]
   ↓    [ 135   85   68 ]
        [ ...           ]
```

````python
print("Mittelwert pro Messgröße (axis=0):", vitalwerte.mean(axis=0))
print("Maximum pro Patient     (axis=1):", vitalwerte.max(axis=1))
````

Ein klinisches Beispiel: der **mittlere arterielle Druck (MAD)** ≈ diastolisch + ⅓ × (systolisch − diastolisch).

````python
systolisch  = vitalwerte[:, 0]
diastolisch = vitalwerte[:, 1]

mad = diastolisch + (systolisch - diastolisch) / 3
print(np.round(mad, 1))
````

## 9. Zufallszahlen: Daten simulieren

Für Übungen, Simulationen oder Stichprobenziehungen brauchen wir oft Zufallszahlen. Der `seed` (hier 42) sorgt dafür, dass bei jedem Ausführen **dieselben** Zufallszahlen entstehen – wichtig für Reproduzierbarkeit!

````python
rng = np.random.default_rng(seed=42)

# 1000 simulierte Körpergrößen: normalverteilt, Mittelwert 172 cm, SD 9 cm
groessen = rng.normal(loc=172, scale=9, size=1000)

print("Erste 5 Werte:", np.round(groessen[:5], 1))
print(f"Mittelwert: {groessen.mean():.1f} cm, SD: {groessen.std(ddof=1):.1f} cm")
````

````python
# Würfeln: ganze Zahlen von 1 bis 6
print(rng.integers(1, 7, size=10))

# Zufällige Zuteilung zu Studienarmen (Randomisierung!)
print(rng.choice(["Placebo", "Verum"], size=8))
````

## 10. Fehlende Werte: `NaN`

In echten Daten fehlen oft Messwerte. NumPy kennzeichnet diese mit `np.nan` (*Not a Number*). Vorsicht: Ein einziger fehlender Wert „vergiftet“ normale Berechnungen.

````python
crp = np.array([3.2, 12.5, np.nan, 0.8, 45.1, np.nan])   # CRP in mg/l

print("np.mean:   ", np.mean(crp))       # -> nan
print("np.nanmean:", np.nanmean(crp))    # ignoriert fehlende Werte
print("Anzahl fehlend:", np.isnan(crp).sum())
````

> In Pandas (nächstes Notebook) ist der Umgang mit fehlenden Werten noch komfortabler.
